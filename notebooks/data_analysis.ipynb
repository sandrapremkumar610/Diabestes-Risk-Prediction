# Diabetes Risk Prediction: Exploratory Data Analysis

This notebook explores the Pima Indians Diabetes Dataset. Run `python train_model.py` first so that `dataset/diabetes.csv` exists.

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# Works whether you start the notebook from the project folder or from notebooks/
path = "../dataset/diabetes.csv" if os.path.exists("../dataset/diabetes.csv") else "dataset/diabetes.csv"
df = pd.read_csv(path)
```

## 1. Dataset overview

```python
print("First five records:")
df.head()
```

```python
print("Shape (rows, columns):", df.shape)
print()
df.info()
```

```python
df.describe().T
```

## 2. Missing-value analysis

In Glucose, BloodPressure, SkinThickness, Insulin and BMI, a zero is not a real measurement, so it is counted as missing.

```python
zero_columns = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"]
print("Missing values stored as NaN:")
print(df.isnull().sum())
print()
print("Duplicate rows:", df.duplicated().sum())
print()
zeros = (df[zero_columns] == 0).sum()
zero_table = pd.DataFrame({"zeros": zeros, "percent": (zeros / len(df) * 100).round(1)})
zero_table
```

```python
# Replace the invalid zeros with NaN for the plots below
clean = df.copy()
clean[zero_columns] = clean[zero_columns].replace(0, np.nan)
clean[zero_columns].isnull().sum()
```

## 3. Outcome distribution

```python
counts = df["Outcome"].value_counts().sort_index()
print(counts.rename({0: "Non-diabetic (0)", 1: "Diabetic (1)"}))

ax = sns.countplot(x="Outcome", data=df)
ax.set_xticks([0, 1])
ax.set_xticklabels(["Non-diabetic (0)", "Diabetic (1)"])
ax.set_title("Outcome distribution")
ax.set_ylabel("Number of records")
plt.show()
```

## 4. Glucose distribution

```python
sns.histplot(clean["Glucose"].dropna(), bins=25, kde=True)
plt.title("Glucose distribution (zeros treated as missing)")
plt.xlabel("Glucose (mg/dL)")
plt.show()
```

## 5. BMI distribution

```python
sns.histplot(clean["BMI"].dropna(), bins=25, kde=True, color="green")
plt.title("BMI distribution (zeros treated as missing)")
plt.xlabel("BMI (kg/m²)")
plt.show()
```

## 6. Age distribution

```python
sns.histplot(df["Age"], bins=20, kde=True, color="purple")
plt.title("Age distribution")
plt.xlabel("Age (years)")
plt.show()
```

## 7. Correlation heatmap

```python
plt.figure(figsize=(9, 7))
sns.heatmap(clean.corr(), annot=True, fmt=".2f", cmap="coolwarm", square=True)
plt.title("Correlation between features and Outcome")
plt.show()
```

## 8. Glucose by outcome

```python
sns.boxplot(x="Outcome", y="Glucose", data=clean)
plt.xticks([0, 1], ["Non-diabetic (0)", "Diabetic (1)"])
plt.title("Glucose level by outcome")
plt.show()
```

## Notes

- The dataset has only Pima Indian women aged 21 and above, so it does not represent every population.
- Many Insulin and SkinThickness values are missing, which is why `train_model.py` uses median imputation inside the pipeline.
- This analysis is to understand the data. It is not a medical study.
